```python
import numpy as np
import pandas as pd
df=pd.read_csv('Property Sales of Melbourne City.csv')
print(df.head())
print(df.describe())
```

```python
range_a=df['City_A'].max()-df['City_A'].min()
range_b=df['City_B'].max()-df['City_B'].min()
print('Range of city A:',range_a)
print('Range of city B:',range_b)
```

```python
var_a=df['City_A'].var()
var_b=df['City_B'].var()
print('Variance of City A:',var_a)
print('Variance of City B:',var_b)
```

```python
std_a=df['City_A'].std()
std_b=df['City_B'].std()
print('Standard Deviation:',std_a)
print('Standard Deviation:',std_b)
```

```python
#coefficient of variation
cv_a = (std_a / df['City_A'].mean()) * 100
cv_b = (std_b / df['City_B'].mean()) * 100
print('CV% - City A:', cv_a)
print('CV% - City B:', cv_b)
```

```python
if std_a < std_b:
    print('City A has more consistent temperatures than City B')
else:
    print('City B has more consistent temperatures than City A')
```
