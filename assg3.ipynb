```python
data=[1,2,2,3,4,5,5,5,6,7]

#Mean
total=0
count=0
for x in data:
    total=total+x
    count=count+1
mean=total/count

#Median
if count%2==1:
    median=data[count//2]
else:
    middle1=data[(count//2)-1]
    middle2=data[count//2]
    median=(middle1+middle2)/2

#Mode
mode=data[0]
max_freq=1
curr_freq=1

for i in range(1,count):
    if data[i]==data[i-1]:
        curr_freq=curr_freq+1
    else:
        curr_freq=1

    if curr_freq>max_freq:
        max_freq=curr_freq
        mode=data[i]
print("Mean:",mean)
print("Mode:",mode)
print("Median:",median)
```

```python
import pandas as pd
df=pd.read_csv("Commonwealth_Medalists_1930_2026.csv")
data=df["Year"]
mean=data.mean()
median=data.median()
mode=data.mode()[0]
print("Mean:",mean)
print("Median:",median)
print("Mode:",mode)
```
